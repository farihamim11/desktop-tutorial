```python
sabcg_graph = {
    'S': [('A', 1), ('B', 1)],
    'A': [('C', 1)],
    'B': [('C', 2)],
    'C': [('G', 3)],
    'G': [],
}
h_sabcg = {'S': 2, 'A': 4, 'B': 1, 'C': 1, 'G': 0}
```

```python
def astar_priority_sabcg(g,node):
  return g+h_sabcg[node]
```

```python
def graph_search(graph, start, goal, priority_fn):
    closed = []
    frontier = [(priority_fn(0, start), 0, start, [start])]
    expansions = 0

    while len(frontier) > 0:
        best_index = 0

        for i in range(len(frontier)):
            if frontier[i][0] < frontier[best_index][0]:
                best_index = i
        priority, g, node, path = frontier.pop(best_index)

        if node == goal:
            return path, g, expansions

        if node not in closed:
            closed.append(node)
            expansions += 1

            for neighbor, weight in graph[node]:
                new_g = g + weight
                new_path = path + [neighbor]

                frontier.append(
                    (priority_fn(new_g, neighbor),
                     new_g,
                     neighbor,
                     new_path)
                )

    return None, None, expansions
```

```python
path, cost, expansions = graph_search(
    sabcg_graph,
    'S',
    'G',
    astar_priority_sabcg
)
print("Fariha")
print("2254901100")
print("Path:", " -> ".join(path) if path else "No path found")
print("Total Cost:", cost)
print("Nodes Expanded:", expansions)
```

2. Trace the code for both search algorithms. Find the value of f (= g + h) for each step. Report if any edges violate admissibility and consistency.

```python
def trace_astar(graph, start, goal):
    closed = []
    frontier = [(bd_heuristics[start], 0, start, [start])]

    print("Node\tg\t h\t f")

    while len(frontier) > 0:
        best_index = 0

        for i in range(len(frontier)):
            if frontier[i][0] < frontier[best_index][0]:
                best_index = i

        f, g, node, path = frontier.pop(best_index)

        if node == goal:
            print(node, "\t", g, "\t", bd_heuristics[node], "\t", f)
            break

        if node not in closed:
            closed.append(node)

            print(node, "\t", g, "\t", bd_heuristics[node], "\t", f)

            for neighbor, weight in graph[node]:
                new_g = g + weight
                new_f = new_g + bd_heuristics[neighbor]

                frontier.append(
                    (new_f,
                     new_g,
                     neighbor,
                     path + [neighbor])
                )
```

```python
trace_astar(
    bd_graph_weighted,
    'Rangpur',
    "Cox's Bazar"
)
```

Trace Explanation:
A* selects the node with the lowest f-value, where f = g + h. It gradually moves from Rangpur toward Cox's Bazar and finally reaches the goal with a total cost of 670.

```python
def check_heuristics(graph, heuristics):
    print("Checking heuristic properties...\n")

    for node in graph:
        for neighbor, cost in graph[node]:
            admissible = heuristics[node] <= cost + heuristics[neighbor]
            consistent = heuristics[node] <= cost + heuristics[neighbor]

            if not admissible:
                print("Admissibility violation:",
                      node, "->", neighbor)

            if not consistent:
                print("Consistency violation:",
                      node, "->", neighbor)
```

```python
check_heuristics(bd_graph_weighted, bd_heuristics)
```

 output explanation:
The heuristic is not fully admissible or consistent because some heuristic values are greater than the actual cost through neighboring nodes. Therefore, A* with this heuristic cannot be guaranteed to be optimal in general.
